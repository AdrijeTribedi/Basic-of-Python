 13.Difference between for and while loop
- i > for loop is used when the number of iteration is known and while loop is used when the number of iteration is unknown.
  ii >for loop is used to iterate the elements of a given sequence,while loop allows to repeat the code as long as the condition is true.
  iii >As the number of iteration is known for loop is faster than while loop,where as while loop is slower than for loop.

```python
n = int(input("Enter the number which have to be reverse : "))
i = 0
temp = n
while(temp>0):
    a = temp % 10
    i = i*10 + a
    temp = temp // 10
print("The reverse of ",n," will be ",i)

```
